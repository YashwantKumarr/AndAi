```python
# ------------------------------------------------------------------
# FULL IMPLEMENTATION: FASTAPI (REST) + MCP (SSE) + NGROK
# ------------------------------------------------------------------
import os
import time
import nest_asyncio
import threading
import uvicorn
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from mcp.server.fastmcp import FastMCP
from pyngrok import ngrok

# 1. Clean up old ports
print("🧹 Cleaning up port 8000...")
os.system("fuser -k 8000/tcp")
time.sleep(2)
nest_asyncio.apply()

# 2. Authenticate Ngrok
NGROK_TOKEN = "3EtlpNjHBEf9AeUTzxXujVkfaeQ_4bFQzsMqJwknNrKa2vp3N"
ngrok.set_auth_token(NGROK_TOKEN)

# =================================================================
# 3. FASTAPI IMPLEMENTATION (The REST API side)
# =================================================================
app = FastAPI(
    title="My Kaggle AI Backend",
    description="A hybrid server serving both REST API requests and MCP tool streaming."
)

# Define a Pydantic Model (FastAPI uses this to validate incoming data)
class AIRequest(BaseModel):
    agent_name: str
    task: str
    urgency: int = 1

# A basic GET route
@app.get("/")
def home():
    return {"message": "FastAPI is running successfully!"}

# A complex POST route using our Pydantic model
@app.post("/submit-task")
def receive_task(payload: AIRequest):
    if payload.urgency > 5:
        raise HTTPException(status_code=400, detail="Urgency cannot exceed 5.")
    
    return {
        "status": "Task Received",
        "assigned_to": payload.agent_name,
        "action": f"Processing task: {payload.task}"
    }

# =================================================================
# 4. MCP IMPLEMENTATION (The AI Tool side)
# =================================================================
mcp = FastMCP("Kaggle-Tools")

@mcp.tool()
async def check_gpu_status() -> str:
    """Checks if Kaggle has a GPU accelerator attached."""
    import torch
    if torch.cuda.is_available():
        return f"GPU is available! Device: {torch.cuda.get_device_name(0)}"
    return "No GPU attached. Running on CPU."

# Mount the MCP server onto the FastAPI app
app.mount("/mcp", mcp.sse_app())

# =================================================================
# 5. START SERVER & TUNNEL
# =================================================================
ngrok.kill() 
public_url = ngrok.connect(8000)

print("-" * 60)
print(f"📖 FASTAPI DOCS UI: {public_url.public_url}/docs")
print(f"🔗 MCP ENDPOINT:    {public_url.public_url}/mcp/sse")
print("-" * 60)

def start_server():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="warning")

threading.Thread(target=start_server, daemon=True).start()
print("🚀 Server is live!")
```

```python
import hmac
import hashlib
from fastapi import FastAPI, Request, Header, HTTPException, Depends
from pydantic import BaseModel

app = FastAPI(title="Secure HMAC API")

# 1. Define your shared secret key (In production, load this from environment variables!)
API_SECRET_KEY = "my_super_secret_production_key".encode("utf-8")

# 2. Create the Verification Dependency
async def verify_hmac_signature(
    request: Request, 
    x_signature: str = Header(..., description="The HMAC-SHA256 signature")
):
    # Step A: Get the raw bytes of the request body
    # We must hash the exact raw bytes, not a parsed JSON dictionary!
    raw_body = await request.body()
    
    # Step B: Calculate what the signature *should* be
    expected_signature = hmac.new(
        key=API_SECRET_KEY,
        msg=raw_body,
        digestmod=hashlib.sha256
    ).hexdigest()
    
    # Step C: Compare the signatures safely
    # We use compare_digest instead of "==" to prevent timing attacks
    if not hmac.compare_digest(expected_signature, x_signature):
        raise HTTPException(status_code=401, detail="Invalid HMAC Signature. Access Denied.")
    
    return True

# 3. Define a normal Pydantic model for your data
class AgentTask(BaseModel):
    agent_id: str
    command: str

# 4. Protect your route by injecting the dependency
@app.post("/execute-task", dependencies=[Depends(verify_hmac_signature)])
async def execute_task(payload: AgentTask):
    return {
        "status": "Success", 
        "message": "Signature verified! Task accepted.",
        "data": payload
    }
```
